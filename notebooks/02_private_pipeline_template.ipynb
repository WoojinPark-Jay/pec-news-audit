## Reader guide: what this template checks

This notebook is a **schema gate** for authorized private reruns. The public repository intentionally excludes raw operational logs, user histories, and row-level processed features. Therefore this notebook does not disclose private data. It checks whether the expected processed tables exist and whether their columns match the pipeline assumptions.

A successful schema check means the environment is ready for the deeper rerun notebooks. It does not by itself prove the paper results; it only confirms that the required private inputs are present.

```python
from pathlib import Path
import os
import sys
import pandas as pd
ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from pecgap.private_schema import PRIVATE_TABLES, missing_columns
processed = ROOT / 'data' / 'processed'
processed
```

```python
for table, required in PRIVATE_TABLES.items():
    path = processed / table
    print('\n', table)
    if not path.exists():
        print('  missing private table')
        continue
    df = pd.read_csv(path, nrows=5)
    missing = missing_columns(table, list(df.columns))
    print('  columns:', len(df.columns))
    print('  missing:', missing or 'none')
```
